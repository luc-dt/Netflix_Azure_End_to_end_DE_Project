# Silver Lookup Table Ingestion

**Purpose:** Process lookup datasets such as cast, directors, countries, and categories from the Bronze layer into the Silver layer using reusable Databricks widgets.

This notebook is parameterized with `sourcefolder` and `targetfolder` so the same logic can run for multiple lookup files in a Databricks Workflow.

### Parameters

```python
dbutils.widgets.text("sourcefolder", "netflix_directors")
dbutils.widgets.text("targetfolder", "netflix_directors")
```

### Variables

```python
var_src_folder = dbutils.widgets.get("sourcefolder")
var_tgt_folder = dbutils.widgets.get("targetfolder")
```

```python
df = spark.read.format("csv")\
    .option("header", True)\
    .option("inferSchema", True)\
    .load(f"abfss://bronze@nextflixprojectdltluc.dfs.core.windows.net/{var_src_folder}")
```

```python
display(df.limit(10))
```

```python
df.write.format("delta")\
        .mode("append")\
        .option("path", f"abfss://silver@nextflixprojectdltluc.dfs.core.windows.net/{var_tgt_folder}")\
        .save()
```
