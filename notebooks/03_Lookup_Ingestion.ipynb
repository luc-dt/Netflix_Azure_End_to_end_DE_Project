# Lookup Ingestion - Workflow Configuration

**Purpose:** Define lookup-table metadata and share it with downstream Databricks Workflow tasks.

This notebook defines the lookup file metadata array and publishes it via Databricks Task Values for downstream workflow tasks. It serves as the **configuration task** in the Databricks Workflow, enabling dynamic ForEach processing of multiple lookup datasets.

### Array Parameter

```python
files = [
    {"sourcefolder": "netflix_directors", "targetfolder": "netflix_directors"},
    {"sourcefolder": "netflix_cast", "targetfolder": "netflix_cast"},
    {"sourcefolder": "netflix_countries", "targetfolder": "netflix_countries"},
    {"sourcefolder": "netflix_category", "targetfolder": "netflix_category"}
]
```

### Job Utility to return the ARRAY

```python
dbutils.jobs.taskValues.set(key="my_arr", value=files)
```
